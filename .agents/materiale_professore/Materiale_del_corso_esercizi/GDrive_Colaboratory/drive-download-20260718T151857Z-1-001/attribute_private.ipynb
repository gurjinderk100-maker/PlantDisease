```python
# p2.py
class P:
   def __init__(self, name="UGO", alias="my data: "):
      self.name = name       # public
      self._protected = alias+"protected"  
      self.__private = alias+"private"   # private

   def who(self):
      print('name  : ', self.name)
      print('alias : ', self.__private)

   def __foo(self):          # private method
      print('This is private method')

   def foo(self):            # public method
      print('This is public method')
      self.__foo()
```

```python
p=P()
print(p.name)
```

```python
print(p._protected)
```

```python
print(p.__private)
```

```python
print(p.foo())
```

```python
print(p.__foo())
```
