```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset (a public dataset provided by seaborn)
iris = sns.load_dataset('iris')
iris
```

```python

# --- Scatter Plot: Petal Length vs Petal Width ---

# Create a scatter plot with different colors for each species
plt.figure(figsize=(8, 6))
species = iris['species'].unique()
colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}

for sp in species:
    subset = iris[iris['species'] == sp]
    plt.scatter(subset['petal_length'], subset['petal_width'],
                color=colors[sp], label=sp, alpha=0.7)

plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Iris Dataset: Petal Length vs Petal Width')
plt.legend()
plt.grid(True)
plt.show()

for sp in species:
    subset = iris[iris['species'] == sp]
    plt.scatter(subset['sepal_length'], subset['sepal_width'],
                color=colors[sp], label=sp, alpha=0.7)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Iris Dataset: Petal Length vs Petal Width')
plt.legend()
plt.grid(True)
plt.show()
```

```python

# --- Histogram: Sepal Length Distribution ---

plt.figure(figsize=(8, 6))
plt.hist(iris['sepal_length'], bins=20, color='purple', alpha=0.7)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Sepal Length in Iris Dataset')
plt.grid(True)
plt.show()
```

```python

# Create a pairplot with species hue
sns.pairplot(iris, hue="species", diag_kind="kde", corner=True)
plt.suptitle('Iris Pairplot', y=1.02)
plt.show()
```

```python
g = sns.PairGrid(iris, hue='species')
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot, cmap="Blues")
g.map_diag(sns.kdeplot)
g.add_legend()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.violinplot(x="species", y="sepal_length", data=iris, palette="muted")
plt.title('Sepal Length Distribution by Species')
plt.show()
```

```python

# Create a boxplot for all numeric features grouped by species
plt.figure(figsize=(8, 6))
sns.boxplot(x="species", y="petal_length", data=iris, palette="Set2")
plt.title('Petal Length Distribution by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Compute the correlation matrix (excluding the species categorical variable)
corr = iris.drop('species', axis=1).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title('Correlation Matrix for Iris Features')
plt.show()
```

```python
from math import pi
import numpy as np

# Select one species (average values for each feature)
species_mean = iris.groupby("species").mean()

# Choose a species for the radar chart (e.g., "setosa")
species = "setosa"
values = species_mean.loc[species].values

# Feature names (axes labels)
categories = list(iris.columns[:-1])
num_vars = len(categories)

# Compute the angle for each axis
angles = np.linspace(0, 2 * pi, num_vars, endpoint=False).tolist()

# Close the loop for radar chart
values = np.concatenate((values, [values[0]]))
angles += angles[:1]

# Create polar plot
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))

# Plot data
ax.plot(angles, values, color='blue', linewidth=2, linestyle='solid')
ax.fill(angles, values, color='blue', alpha=0.25)

# Labels
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)

# Title
plt.title(f'Radar Chart for {species}')
plt.show()
```
