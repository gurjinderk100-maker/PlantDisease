For the following vectors, x and y, calculate the indicated similarity or distance measures.

(a) x = (1, 1, 1, 1), y = (2, 2, 2, 2) cosine, correlation, Euclidean
(b) x = (0, 1, 0, 1), y = (1, 0, 1, 0) cosine, correlation, Euclidean, Jaccard
(c) x = (0, −1, 0, 1), y = (1, 0, −1, 0) cosine, correlation, Euclidean
(d) x = (1, 1, 0, 1, 0, 1), y = (1, 1, 1, 0, 0, 1) cosine, correlation, Jaccard
(e) x = (2, −1, 0, 2, 0, −3), y = (−1, 1, −1, 0, 0, −1) cosine, correlation

${\text{similarity}}=\cos(\theta )={A\cdot B \over \|A\|\|B\|}.$

l'indice di correlazione di Pearson è definito come la loro covarianza divisa per il prodotto delle deviazioni standard delle due variabili:
${\displaystyle \rho _{XY}={\frac {\sigma _{XY}}{\sigma _{X}\sigma _{Y}}}.}$
dove ${\displaystyle \sigma _{XY}}$ è la covarianza tra ${\displaystyle X}$ e ${\displaystyle Y}$ e ${\displaystyle \sigma _{X},\sigma _{Y}}$ sono le due deviazioni standard.


Jaccard: ${\displaystyle J={M_{11} \over M_{01}+M_{10}+M_{11}}.}$

```python
import numpy as np
from sklearn.metrics import jaccard_score
from scipy.spatial import distance
from scipy.spatial import minkowski_distance_p
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import pairwise_distances
from scipy.stats import pearsonr

```

```python
def compute(x,y):
  corr, _ = pearsonr(x,y)
  xm=np.mean(x)
  ym=np.mean(y)

  print("x: ",x," std:", np.std(x), " mean:",xm, " xnorm': ",(x-xm)/np.std(x))
  print("y: ",y," std:", np.std(y), " mean:",ym, " ynorm: ",(y-ym)/np.std(y))
  print("Manhattan distance: " + str(distance.cityblock(x, y)))
  print("Euclidean distance: " + str(distance.euclidean(x, y)))
  print("Chebyshev distance: " + str(distance.chebyshev(x, y)))
  print("-------------------------------------------")
  print("cosine: " + str(cosine_similarity(x.reshape(1,len(x)),y.reshape(1,len(y)))))
  print("correlation: " + str(corr))
  print("jaccard: " + str(jaccard_score(x, y)))
```

```python
#A
x = np.array([1, -1, 1])
y = np.array([-1,1,1])

compute(x,y)
print("jaccardMin: " + str(jaccard_score(x, y, pos_label=-1)))
print("jaccarMax: " + str(jaccard_score(x, y, pos_label=1)))
```

```python
xm=np.mean(x)
xstd=np.std(x)
xs=(x-xm)/xstd
xm,xstd,xs
```

```python
ym=np.mean(y)
ystd=np.std(y)
ys=(y-ym)/ystd
ym,ystd,ys
```

```python
#B
x = np.array([0,1, 1, 0])
y = np.array([-4,-4, 0, 0])

compute(x,y)
print("jaccardMin: " + str(jaccard_score(x, y, pos_label=-1)))
print("jaccarMax: " + str(jaccard_score(x, y, pos_label=1)))
```
